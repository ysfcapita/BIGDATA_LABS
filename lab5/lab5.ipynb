```python
# Installation PySpark
!pip install pyspark

# SparkContext + StreamingContext (batch = 5s)
import random
from pyspark import SparkContext
from pyspark.streaming import StreamingContext

sc = SparkContext.getOrCreate()
ssc = StreamingContext(sc, 5)

# lire le fichier messages.txt (lecture aléatoire)
def read_random_file(_):
    with open("/messages.txt", "r", encoding="utf-8") as f:
        lines = f.readlines()
    return [random.choice(lines).strip()]

# DStream simulé à partir d’un fichier
lines = ssc.queueStream(
    [sc.parallelize(read_random_file(None)) for _ in range(10)]
)

# WordCount streaming
words = lines.flatMap(lambda line: line.split(" "))
pairs = words.map(lambda w: (w, 1))
counts = pairs.reduceByKey(lambda a, b: a + b)

# Affichage dans la console
counts.pprint()

# Lancer le streaming
ssc.start()
ssc.awaitTerminationOrTimeout(30)
ssc.stop(stopSparkContext=False)
```
